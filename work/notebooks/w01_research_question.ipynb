## 2. The question: decision, action, cost of a wrong call

**Research question:** Can observable content and search-performance signals help rank pages for editorial review more usefully than a simple hand-written rule?

- **Decision:** Which pages should an editor review first?
- **Who acts and what they do:** An SEO/content editor reviews the highest-priority pages, checks the evidence, and decides whether an update is warranted. The score is a triage aid, not an automatic instruction to edit.
- **Cost of a wrong call:** A false positive wastes review time on a page that does not need attention; a false negative may leave a genuinely declining page unreviewed. I will compare these costs and avoid treating every flagged page as an urgent fix.
- **Unit of analysis:** One pseudonymized content item/page in the starter dataset.
- **Planned output:** A ranked review queue with a score and understandable reasons for prioritization.
- **Why data/ML might help:** Several observed signals may interact in ways a single threshold misses. A transparent rule is the baseline; a more complex model is justified only if evaluation shows a meaningful improvement over that baseline.

## 3. Quick look at the data (2–3 real numbers)

The starter dataset is published in the official FlyRank starter repository. The code below reads it directly from the repository so the CSV does not need to be copied into this public assignment repository. It prints the row and column counts and checks how many rows have `avg_position == 0` (documented as a no-data value, not a rank of zero). Run this cell to produce the numerical evidence in this notebook.

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(DATA_URL)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns):,}")
if "avg_position" in df.columns:
    no_position = int((df["avg_position"] == 0).sum())
    print(f"Rows with avg_position == 0 (no data, not rank zero): {no_position:,}")
else:
    print("avg_position column not found; inspect df.columns and the data dictionary.")

print("\nColumns available for framing:")
print(", ".join(df.columns))
```
